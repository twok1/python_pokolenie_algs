# Количество цифр в числе

✅ Количество цифр в произвольном положительном целом числе равняется десятичному логарифму этого числа, округлённому до целого в меньшую сторону и увеличенному на единицу:
$$\lfloor \log_{10} n \rfloor + 1$$

✅ Количество цифр произвольного положительного целого числа в его двоичном представлении равняется двоичному логарифму этого числа, округлённому до целого в меньшую сторону и увеличенному на единицу:
$$\lfloor \log_{2} n \rfloor + 1$$

✅ Количество цифр в положительном целом числе $n$ в $b$-ичной системе счисления равняется:
$$\lfloor \log_{b} n \rfloor + 1$$

✅ Запись $\lfloor x \rfloor$ обозначает округление числа $x$ до ближайшего целого в меньшую сторону.
✅ Запись $\lceil x \rceil$ обозначает округление числа $x$ до ближайшего целого в большую сторону.
✅ Для чисел типа `int` существует метод `bit_length()`, возвращающий количество цифр в двоичной записи числа.

✅ **Алгоритм** – это последовательность действий, которые нужно выполнить, чтобы решить определенную задачу.\
✅ **Исполнитель** – это человек, компьютер или любой другой объект, способный выполнить алгоритм.\
✅ **Команда** – это указание в записи алгоритма, которое обязывает исполнителя произвести какое-либо действие

✅ `Свойства алгоритма`:\
Дискретность. Выполнение алгоритма разбивается на последовательность определенных действий. Только выполнив одно действие, можно приступать к исполнению следующего
Понятность. Алгоритм не должен содержать команд, смысл которых может восприниматься исполнителем неоднозначно. Запись алгоритма должна быть настолько четкой, чтобы исполнителю не приходилось принимать решения самостоятельно
Детерминированность. Способ решения задачи однозначно определен в виде последовательности шагов. То есть, если алгоритм многократно применяется к одному и тому же набору исходных данных, то каждый раз получаются одни и те же промежуточные результаты и один и тот же выходной результат
Конечность. При точном исполнении команд алгоритма процесс должен прекратиться за конечное число шагов, и при этом должен быть получен ответ на вопрос задачи. Одним из возможных ответов может быть и установление факта, что задача не имеет решений
Массовость. Алгоритм должен решать не одну частную задачу, а класс задач. Например, не имеет смысла строить алгоритм поиска максимального числа в списке, состоящем только из десяти элементов

✅ `Способы описания алгоритма`:\
**Словесный способ** используется для описания алгоритма на естественном языке в виде пронумерованных строк, соответствующих шагам выполнения алгоритма\
При **графическом способе** алгоритм изображается в виде последовательности связанных между собой блоков, каждый из которых соответствует выполнению определенного действия. Такое графическое представление называется **блок-схемой**.\
**Программный способ** описания алгоритма предполагает написание программы на каком-либо языке программирования с полным соблюдением его синтаксиса\
**Псевдокод** – неформальный язык описания алгоритмов, использующий ключевые слова различных языков программирования, но опускающий несущественные для понимания алгоритма подробности и специфический синтаксис

✅ `Виды алгоритмов`:\
**Последовательный алгоритм** – это алгоритм, состоящий из действий, которые выполняются одно за другим в строгом порядке.\
**Ветвящийся алгоритм** – это алгоритм, который предусматривает выполнение хотя бы одной операции по проверке условия, в результате чего выполняется либо одно действие, либо другое\
**Циклический алгоритм** – это алгоритм, который предусматривает многократное повторение определенной последовательности действий. Количество повторений обычно задается определенным условием

✅ `Этапы проектирования решения алгоритмической задачи`:\
**Этап 1**. Понимание задачи\
**Этап 2**. Разработка алгоритма\
**Этап 3**. Оценка эффективности алгоритма\
**Этап 4**. Реализация алгоритма

✅ `Оценка эффективности алгоритма включает в себя анализ двух параметров`:\
**временнáя эффективность** – это оценка времени работы алгоритма\
**пространственная эффективность** – это оценка количества оперативной памяти, необходимой для работы алгоритма

# Правила учёта операций в циклах

✅ Если на каждой итерации **внешнего цикла** выполняются **простейшие операции**, количество которых **фиксировано и не зависит** от размера входных данных, то эти операции можно не учитывать.

✅ Если на каждой итерации **внутреннего цикла** выполняются **простейшие операции**, количество которых **фиксировано и не зависит от** размера входных данных, то эти операции можно не учитывать.

✅ Если во **внешнем и внутреннем циклах нет операций, количество которых зависит от размера входных данных** (конечно, не учитывая наличие внутреннего цикла в теле внешнего цикла), то при определении временной сложности алгоритма достаточно учитывать только то, чему равняется произведение количества итераций внешнего и внутреннего циклов. Это правило справедливо для любой пары циклов:

- к какому бы типу ни относились циклы (`for` или `while`), правилом пользоваться можно;
- какое бы число итераций ни выполняли циклы (зависящее от размера входных данных или фиксированное и не зависящее от размера входных данных), правилом также можно пользоваться.

✅ Если алгоритм содержит цикл, внутри которого находится другой цикл, а внутри него — ещё один, и так любое количество раз, то при определении временной сложности алгоритма достаточно учитывать только то, чему равняется произведение количества итераций всех этих циклов, при условии, что ни в одном из циклов нет простейших операций, количество которых зависит от размера входных данных.

# Прогрессии

## Арифметическая прогрессия

✅ **Арифметическая прогрессия** — это последовательность чисел $a_1, a_2, a_3, a_4, \ldots$, в которой каждое число, начиная со второго, получается из предыдущего добавлением к нему постоянного значения $d$. Числа $a_1, a_2, a_3, a_4, \ldots$ называются **членами прогрессии**, а значение $d$ — **разностью** или **шагом** прогрессии.

✅ $n$-й член арифметической прогрессии:

$$a_n = a_1 + (n - 1) \cdot d$$

где $a_1$ — первый член прогрессии, а $d$ — шаг прогрессии.

✅ Сумма первых $n$ членов арифметической прогрессии:

$$S_n = \frac{a_1 + a_n}{2} \cdot n$$

или

$$S_n = \frac{2a_1 + (n - 1) \cdot d}{2} \cdot n$$

где $a_1$ — первый член прогрессии, $a_n$ — $n$-й член прогрессии, а $d$ — шаг прогрессии.

## Геометрическая прогрессия

✅ **Геометрическая прогрессия** — это последовательность чисел $a_1, a_2, a_3, a_4, \ldots$, в которой каждое число, начиная со второго, получается из предыдущего числа умножением его на фиксированное значение $q$. Числа $a_1, a_2, a_3, a_4, \ldots$ называются **членами прогрессии**, а значение $q$ — **знаменателем** прогрессии.

✅ $n$-й член геометрической прогрессии:

$$a_n = a_1 \cdot q^{n-1}$$

где $a_1$ — первый член прогрессии, а $q$ — знаменатель прогрессии.

✅ Сумма первых $n$ членов геометрической прогрессии:

$$S_n = \frac{a_1 \cdot (q^n - 1)}{q - 1}$$

где $a_1$ — первый член прогрессии, а $q$ — знаменатель прогрессии.

✅ **Пространственная сложность** алгоритма описывает то, как изменяется **объем памяти**, необходимый для выполнения алгоритма, по мере роста размера входных данных.

✅ **Объем памяти**, используемый алгоритмом во время его выполнения, состоит из трех частей:\
**Память для входных данных** используется для хранения информации, подаваемой на вход алгоритму\
**Дополнительная память** используется для хранения данных, возникающих в процессе выполнения алгоритма\
**Память для выходных данных** используется для хранения результатов работы алгоритма\
✅ **Пространственная сложность** алгоритма определяется точно так же, как и его временная сложность. Сначала составляется функция **S(n)**, которая размеру входных данных **n** сопоставляет максимальный объем памяти (например, в байтах), необходимый для выполнения алгоритма. Затем вычисляется верхняя граница полученной функции с помощью нотации O-большое.

✅ **Пространственная сложность** алгоритма равна O(1), если **объем памяти**, необходимый для выполнения алгоритма, **фиксирован** и не зависит от размера входных данных.


# Анализ алгоритмов с помощью нотации O

✅ Анализ алгоритмов с помощью нотации $O$-большое называется **асимптотическим анализом**. Асимптотический анализ включает оценку как временной сложности алгоритма, так и пространственной.

## Плюсы асимптотического анализа

- Асимптотический анализ упрощает процесс оценки сложности алгоритма, позволяя не учитывать время и память, требуемые для каждой операции внутри алгоритма.
- Оценка сложности алгоритма с помощью асимптотического анализа остаётся актуальной независимо от платформы, на которой выполняется алгоритм.
- Благодаря асимптотическому анализу сложность алгоритмов можно записать в простой форме.
- Асимптотический анализ упрощает сравнение алгоритмов между собой. С его помощью практически всегда можно определить, какой алгоритм будет быстрее или потребует меньше памяти.

## Минусы асимптотического анализа

- Асимптотический анализ не позволяет сравнивать между собой алгоритмы, имеющие одинаковую временную или пространственную сложность.
- Асимптотический анализ может приводить к ошибочным выводам, если постоянные множители, скрываемые нотацией $O$-большое, достаточно велики.

✅ При сравнении алгоритмов бывает важно учитывать не только их итоговые сложности, но и значения постоянных слагаемых и множителей. Это особенно актуально при выборе алгоритма для практического применения.

✅ Современные платформы в среднем способны совершать от $10^6$ до $10^8$ простейших операций в секунду. Это значение не постоянно, так как зависит от множества факторов: мощности процессора, языка программирования и даже от типа простейших операций (например, сложение двух чисел выполняется быстрее, чем их деление).

✅ Таблица-памятка, которая позволяет определять, какую временную сложность должен иметь решающий задачу алгоритм по заданным ограничениям размера входных данных и времени выполнения в 2 секунды:

| Размер входных данных $n$ | Допустимая временная сложность |
|---|---|
| $n \le 10$ | $O(n!)$ |
| $n \le 20$ | $O(2^n)$ |
| $n \le 10^2$ | $O(n^3)$ |
| $n \le 10^3$ | $O(n^2)$ |
| $n \le 10^6$ | $O(n \log n)$ |
| $n \le 10^7$ | $O(n)$ |
| $n > 10^7$ | $O(\log n)$ |

# Сложность встроенных функций
**len()**\
Временная сложность функции: O(1)\
Пространственная сложность функции: O(1)

**min(), max(), sum()**:
 - **В первом варианте** (iterable, key/sum:start, min|max:default):
   - Временная сложность: O(n)
   - Пространственная сложность: O(1)
 - **Во втором варианте** (позиционные аргументы, optional[key]):
   - Временная сложность: O(1)
   - Пространственная сложность: O(1)

**sorted()**:
 - Временная сложность: O(nlog(n))
 - Пространственная сложность: O(n)

**pow()**: 
 - Временная сложность функции: O(log(n))
 - Пространственная сложность функции: O(1)

# Временная и пространственная сложности операций

## Списки (`list`)

Переменная $n$ в сложности операций характеризует длину списка `lst`, а переменная $k$ — длину объекта `iterable`.

| Операция | Временная сложность | Пространственная сложность |
|---|---|---|
| `lst[index]` | $O(1)$ | $O(1)$ |
| `lst[index] = value` | $O(1)$ | $O(1)$ |
| `del lst[index]` | $O(n)$ | $O(1)$ |
| `lst.append(value)` | $O(1)$ | $O(1)$ |
| `lst.insert(index, value)` | $O(n)$ | $O(1)$ |
| `lst.pop()` | $O(1)$ | $O(1)$ |
| `lst.pop(index)` | $O(n)$ | $O(1)$ |
| `value in lst` | $O(n)$ | $O(1)$ |
| `for elem in lst:` | $O(n)$ | $O(1)$ |
| `lst.index(value)` | $O(n)$ | $O(1)$ |
| `lst.remove(value)` | $O(n)$ | $O(1)$ |
| `lst.count(value)` | $O(n)$ | $O(1)$ |
| `lst.sort()` | $O(n \log n)$ | $O(n)$ |
| `lst.reverse()` | $O(n)$ | $O(1)$ |
| `lst.copy()` | $O(n)$ | $O(n)$ |
| `lst.clear()` | $O(n)$ | $O(1)$ |
| `lst.extend(iterable)` | $O(k)$ | $O(k)$ |

✅ Некоторые вызовы метода `append()` могут иметь временную сложность $O(n)$, где $n$ — длина списка.

## Кортежи (`tuple`)

Переменная $n$ в сложности операций характеризует длину кортежа `tpl`.

| Операция | Временная сложность | Пространственная сложность |
|---|---|---|
| `tpl[index]` | $O(1)$ | $O(1)$ |
| `value in tpl` | $O(n)$ | $O(1)$ |
| `for elem in tpl:` | $O(n)$ | $O(1)$ |
| `tpl.index(value)` | $O(n)$ | $O(1)$ |
| `tpl.count(value)` | $O(n)$ | $O(1)$ |

## Объекты `range`

Переменная $n$ в сложности операций характеризует длину объекта `rng`.

| Операция | Временная сложность | Пространственная сложность |
|---|---|---|
| `rng[index]` | $O(1)$ | $O(1)$ |
| `value in rng` | $O(1)$ ⚡ | $O(1)$ |
| `for elem in rng:` | $O(n)$ | $O(1)$ |
| `rng.index(value)` | $O(1)$ ⚡ | $O(1)$ |
| `rng.count(value)` | $O(1)$ ⚡ | $O(1)$ |
| `len(rng)` | $O(1)$ | $O(1)$ |
| `min(rng)` | $O(n)$ | $O(1)$ |
| `max(rng)` | $O(n)$ | $O(1)$ |
| `sum(rng)` | $O(n)$ | $O(1)$ |

## Строки (`str`)

Переменная $n$ в сложности операций характеризует длину строки `s`.

| Операция | Временная сложность | Пространственная сложность |
|---|---|---|
| `s[index]` | $O(1)$ | $O(1)$ |
| `for char in s:` | $O(n)$ | $O(1)$ |
| `s.capitalize()` | $O(n)$ | $O(n)$ |
| `s.lower()` | $O(n)$ | $O(n)$ |
| `s.strip(value)` | $O(n)$ | $O(n)$ |
| `s.isalpha()` | $O(n)$ | $O(1)$ |
| `s.split(value)` | $O(n)$ | $O(n)$ |

## Общие замечания

✅ В основе типов данных `list`, `tuple` и `str` лежит массив.

✅ Операция создания списка, кортежа или строки длины $n$ имеет временную и пространственную сложность $O(n)$.

✅ Операция слияния двух списков вида `lst1 + lst2`, в результате которой создаётся новый список, имеет временную и пространственную сложность $O(n + m)$, где $n$ — длина списка `lst1`, а $m$ — длина списка `lst2`. Операция слияния с изменением левого списка `lst1 += lst2` имеет более низкую временную и пространственную сложность — $O(m)$.

✅ Операции слияния двух кортежей вида `tpl1 + tpl2` и `tpl1 += tpl2` имеют одинаковую временную и пространственную сложность $O(n + m)$, где $n$ — длина кортежа `tpl1`, а $m$ — длина кортежа `tpl2`.

✅ Конкатенация двух строк, выполняемая в виде операции `s1 + s2` или `s1 += s2`, имеет временную и пространственную сложность $O(n + m)$, где $n$ — длина строки `s1`, а $m$ — длина строки `s2`.

✅ Временную сложность операции поиска подстроки в строке с помощью оператора `in` и методов `find()`, `rfind()`, `index()`, `rindex()`, `count()`, `replace()` принято считать равной $O(n)$, где $n$ — длина строки.

## Словари (`dict`)

Переменная $n$ в сложности операций характеризует длину словаря `dct`, а переменная $k$ — длину объекта `iterable`.

| Операция | Временная сложность | Пространственная сложность |
|---|---|---|
| `dct[key]` | $O(1)$ ⚡ | $O(1)$ |
| `dct[key] = value` | $O(1)$ ⚡ | $O(1)$ |
| `del dct[key]` | $O(1)$ ⚡ | $O(1)$ |
| `dct.get(key)` | $O(1)$ ⚡ | $O(1)$ |
| `dct.pop(key)` | $O(1)$ ⚡ | $O(1)$ |
| `key in dct` | $O(1)$ ⚡ | $O(1)$ |
| `value in dct.values()` | $O(n)$ 🐢 | $O(1)$ |
| `for key in dct:` | $O(n)$ | $O(1)$ |
| `for key, value in dct.items():` | $O(n)$ | $O(1)$ |
| `dct.copy()` | $O(n)$ | $O(n)$ |
| `dct.clear()` | $O(n)$ | $O(1)$ |
| `dct.update(iterable)` | $O(k)$ | $O(k)$ |

## Множества (`set`)

Переменная $n$ в сложности операций характеризует длину множеств `s` и `s1`, а переменная $m$ — длину множества `s2`.

| Операция | Временная сложность | Пространственная сложность |
|---|---|---|
| `s.add(value)` | $O(1)$ ⚡ | $O(1)$ |
| `s.remove(value)` | $O(1)$ ⚡ | $O(1)$ |
| `s.discard(value)` | $O(1)$ ⚡ | $O(1)$ |
| `s.pop()` | $O(1)$ ⚡ | $O(1)$ |
| `value in s` | $O(1)$ ⚡ | $O(1)$ |
| `for elem in s:` | $O(n)$ | $O(1)$ |
| `s.copy()` | $O(n)$ | $O(n)$ |
| `s.clear()` | $O(n)$ | $O(1)$ |
| `s1 == s2` | $O(n)$ | $O(1)$ |
| `s1 != s2` | $O(n)$ | $O(1)$ |
| `s1 < s2` | $O(n)$ | $O(1)$ |
| `s1 > s2` | $O(m)$ | $O(1)$ |
| `s1 <= s2` или `s1.issubset(s2)` | $O(n)$ | $O(1)$ |
| `s1 >= s2` или `s2.issubset(s1)` | $O(m)$ | $O(1)$ |
| `s1 \| s2` или `s1.union(s2)` | $O(n + m)$ | $O(n + m)$ |
| `s1 & s2` или `s1.intersection(s2)` | $O(\min(n, m))$ | $O(\min(n, m))$ |
| `s1 - s2` или `s1.difference(s2)` | $O(n)$ | $O(n)$ |
| `s1 ^ s2` или `s1.symmetric_difference(s2)` | $O(n + m)$ | $O(n + m)$ |

## Общие замечания

✅ В основе типов данных `dict` и `set` лежит хеш-таблица.

✅ Операция создания словаря или множества длины $n$ имеет временную и пространственную сложность $O(n)$.
